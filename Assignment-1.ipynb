## 1. Importing Library

```python
import pandas as pd
```

```python
#load the dataset
df = pd.read_csv('pandas_dataset.csv')
initial_shape = df.shape

print("Shape before cleaning:", initial_shape)
print("\n")
print("First Five Rows")
print(df.head())
```

## 2. Renaming Columns

```python
# Cleaning up column names so they're easier to work with

df.columns = df.columns.str.lower().str.replace(' ', '_')

print("Cleaned Column Names:")
print(df.columns.tolist())
```

## 3. Dropping Unnecessary Rows

```python
# First row after the header is blank, get rid of it

df = df.iloc[1:]
df = df.reset_index(drop=True)

print("First 5 Rows (Cleaned):")
print(df.head())
```

## 4. Dropping Unecessary Columns

```python
# city isn't used in any filtering, grouping, or new-column step below, so it's safe to drop

df = df.drop('city', axis=1, errors='ignore')

print("Remaining Columns:")
print(df.columns.tolist())
```

## 5. Counting Missing Values

```python
print("No. of Null values per column:")
print(df.isnull().sum())
```

## 6. Handling Missing Values

**a. Drop Rows Based on a Column**

```python
df = df.dropna(subset=['sales'])

print("Shape after dropping rows with missing sales:")
print(df.shape)
```

**b. Fill Missing Values**

```python
# Profit gets the mean, discount gets the median (less sensitive to outliers)
df['profit'] = df['profit'].fillna(df['profit'].mean())
df['discount'] = df['discount'].fillna(df['discount'].median())

print("Null values remaining in profit and discount:")
print(df[['profit', 'discount']].isnull().sum())
```

## 7. Handling Missing Categorical Data

```python
df['customer_name'] = df['customer_name'].fillna('Unknown')

print("Null values remaining in customer_name:")
print(df['customer_name'].isnull().sum())
```

## 8. Detecting and Removing Duplicates

```python
# checking duplicate order_id values first
print("Duplicate order_id count:")
print(df.duplicated(subset=['order_id']).sum())
```

```python
rows_before = df.shape[0]
df = df.drop_duplicates(subset=['order_id'])

print("Removed:", rows_before - df.shape[0])
```

## 9. Filtering and creating new columns

```python
print("Orders where unit_price > 20,000:")
print(df[df['unit_price'] > 20000])
```

```python
print("Orders where unit_price > 10,000 and status is Completed:")
print(df[(df['unit_price'] > 10000) & (df['status'] == 'Completed')][['customer_name', 'category', 'status']])
```

```python
# total_amount is quantity times price, minus the discount
df['total_amount'] = df['quantity'] * df['unit_price'] - df['discount']

print(df.head())
```

```python
df['customer_type'] = df['quantity'].apply(lambda x: 'Bulk Buyer' if x >= 3 else 'Regular Buyer')

print(df.head())
```

## 9. Final Clean Dataset

```python
print("Before:", initial_shape)
print("After:", df.shape)
```
